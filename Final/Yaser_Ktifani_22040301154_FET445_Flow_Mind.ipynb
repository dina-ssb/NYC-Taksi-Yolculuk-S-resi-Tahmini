## Train Random Forest Regressor

### Subtask:
Drop non-numeric and irrelevant columns, convert categorical features to numeric, and train a RandomForestRegressor model on a sample of the preprocessed training data.


**Reasoning**:
The subtask involves several data preprocessing steps and then training a RandomForestRegressor. I will start by dropping specified columns, converting categorical features, sampling the data, and then training the model, as instructed.



```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# 1. Define columns to drop
columns_to_drop = ['id', 'pickup_datetime', 'dropoff_datetime']

# 2. Create new DataFrames by dropping these columns
X_train_model = X_train_processed.drop(columns=columns_to_drop)
X_test_model = X_test_processed.drop(columns=columns_to_drop)

print(f"X_train_model shape after dropping columns: {X_train_model.shape}")
print(f"X_test_model shape after dropping columns: {X_test_model.shape}")

# 3. Convert 'store_and_fwd_flag' from categorical to numerical
X_train_model['store_and_fwd_flag'] = X_train_model['store_and_fwd_flag'].map({'N': 0, 'Y': 1})
X_test_model['store_and_fwd_flag'] = X_test_model['store_and_fwd_flag'].map({'N': 0, 'Y': 1})

print("Converted 'store_and_fwd_flag' to numerical.")
print("First 5 rows of X_train_model after conversion:")
display(X_train_model.head())

# 5. Instantiate a RandomForestRegressor model
# Using n_estimators=100 as a starting point, random_state for reproducibility, n_jobs=-1 for using all processors
model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
print("RandomForestRegressor model instantiated.")

# 6. Select a sample of the training data
sample_fraction = 0.1 # Using 10% of the training data for sampling
X_train_sample = X_train_model.sample(frac=sample_fraction, random_state=42)
y_train_sample = y_train.loc[X_train_sample.index]

print(f"Created training sample with {len(X_train_sample)} records ({sample_fraction*100}% of original training data).")

# 7. Train the RandomForestRegressor model
print("Training RandomForestRegressor model on sampled data...")
model.fit(X_train_sample, y_train_sample)
print("RandomForestRegressor model training complete.")
```

## Train XGBoost Regressor

### Subtask:
Train an XGBoostRegressor model using the identified features (X_train_processed) and target variable (y_train), similar to how the Random Forest was trained.


**Reasoning**:
The subtask is to train an XGBoostRegressor model. This involves importing the necessary class, instantiating the model with specified parameters, and then fitting it to the `X_train_sample` and `y_train_sample` data.



```python
import xgboost as xgb
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error

# Re-load df_from_drive and split data (as it seems the state was reset or not fully executed)
file_path = '/content/drive/MyDrive/train_clean_final.csv'
df_from_drive = pd.read_csv(file_path)

X_train, X_test = train_test_split(df_from_drive, test_size=0.2, random_state=42)

target_column = 'trip_duration'

y_train = X_train[target_column]
X_train_processed = X_train.drop(target_column, axis=1)

y_test = X_test[target_column]
X_test_processed = X_test.drop(target_column, axis=1)

# Re-define columns to drop and create X_train_model, X_test_model (from previous steps)
columns_to_drop = ['id', 'pickup_datetime', 'dropoff_datetime']

X_train_model = X_train_processed.drop(columns=columns_to_drop)
X_test_model = X_test_processed.drop(columns=columns_to_drop)

# Re-convert 'store_and_fwd_flag' to numerical
X_train_model['store_and_fwd_flag'] = X_train_model['store_and_fwd_flag'].map({'N': 0, 'Y': 1})
X_test_model['store_and_fwd_flag'] = X_test_model['store_and_fwd_flag'].map({'N': 0, 'Y': 1})

# 1. Re-create the sample to ensure data consistency
sample_fraction = 0.1
X_train_sample = X_train_model.sample(frac=sample_fraction, random_state=42)
y_train_sample = y_train.loc[X_train_sample.index]

print(f"Re-created training sample with {len(X_train_sample)} records.")

# 2. Instantiate the XGBoost model
xgb_model = xgb.XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42,
    tree_method='hist',  # Changed from 'gpu_hist' to 'hist' for compatibility
    n_jobs=-1
)

# 3. Train the model
print("Training XGBoostRegressor model on sampled data...")
xgb_model.fit(X_train_sample, y_train_sample)
print("XGBoostRegressor model training complete.")

# 4. Calculate performance metrics
y_pred_xgb = xgb_model.predict(X_test_model)
r2 = r2_score(y_test, y_pred_xgb)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_xgb))

print("-" * 30)
print(f"XGBoost Final Results:")
print(f"R-squared Score: {r2:.4f}")
print(f"RMSE: {rmse:.2f}")
print("-" * 30)

# 5. Plot Feature Importance
import matplotlib.pyplot as plt
xgb.plot_importance(xgb_model, max_num_features=10, importance_type='weight')
plt.title('XGBoost: Top 10 Features Influencing Trip Duration')
plt.show()
```
